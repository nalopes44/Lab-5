# Verifying relationships

Here we are verifying each child class is related to it's parent class for weapons and characters.

```python
print("Fighter inherits from Character:", issubclass(Fighter, Character))
print("Mage inherits from Character:", issubclass(Mage, Character))

print("MeleeWeapon inherits from Weapon:", issubclass(MeleeWeapon, Weapon))
print("RangedWeapon inherits from Weapon:", issubclass(RangedWeapon, Weapon))
print("MagicWeapon inherits from Weapon:", issubclass(MagicWeapon, Weapon))
```

# Test A: Valid melee attack
Testing whether a valid melee attack registers and reduces the targets health.

We expect the mage to have a health of 65 after the attack

```python
fighter = Fighter("Arthur", 100, 10)
target = Mage("Enemy mage", 80, 5)
sword = MeleeWeapon("Sword", 15)

health_before = target.health
attack(fighter, sword, target, distance=0)

print("Expected target health:", 65)
print("Actual target health:", target.health)

assert target.health == 65
print("Test passed: Target health is correct after attack.")
```

## Test B: Ranged weapon attack

this test checks if a ranged weapon can attack a target within its allowed range

Arthur's health decreases from 80 to 68

```python
bow = RangedWeapon("Bow", 12, 10)
target = Mage("Enemy mage", 80, 5)

health_before = target.health
attack(fighter, bow, target, distance=5)

print("Health before attack:", health_before)
print("Expected health after attack:", 68)

print("Actual health after attack:", target.health)
print("Damage received:", health_before - target.health)

assert target.health == 68
print("Test passed: Target health is correct after ranged attack.")

```

# Test C: Valid magic attack

We are testing whether a valid magic attack will reduce the health of the target

After the attack, we expect the target health to be 80, down from 100

```python
staff = MagicWeapon("Staff", 20, 10, 5)
mage = Mage("Gandalf", 80, 5)
target = Fighter("Enemy fighter", 100, 8)

attack(mage, staff, target, distance=5)

print("Expected target health:", 80)
print("Actual target health:", target.health)

assert target.health == 80
print("Test passed: Target health is correct after magic attack.")
```

# Special cases

Finally, we are displaying a special case where a characters health is reduced below 0. In this case, the health is set to 0.

```python
bow = RangedWeapon("Bow", 12, 10)
target = Fighter("Target", 50, 8)

attack(fighter, bow, target, distance=11)

print("Expected target health:", 38)
print("Actual target health:", target.health)
assert target.health == 38

#test 2: target health must not become negative
strong_sword = MeleeWeapon("Strong Sword", 100)
target = Fighter("Target", 20, 5)

attack(fighter, strong_sword, target, distance=0)

print("Expected target health:", 0)
print("Actual target health:", target.health)
assert target.health == 0

print("Test passed: Target health is correct after strong attack, not negative.")
```
